```python
# Örnek veri setleri
data1 = {
    "ID": [1, 2, 3],
    "Type": ["R", "M", "R"],
    "Message": ["Hello", "Hi", "How are you?"],
    "From": ["agent1", "john", "agent1"]
}

data2 = {
    "ID": [4, 5, 6],
    "Type": ["M", "R", "M"],
    "Message": ["Hey", "Hello", "I'm good"],
    "From": ["emily", "agent1", "emily"]
}

data3 = {
    "ID": [7, 8, 9],
    "Type": ["R", "R", "M"],
    "Message": ["Hi there", "What's up?", "Goodbye"],
    "From": ["agent1", "agent1", "alice"]
}

# Veri setlerini DataFrame'e dönüştürme
df1 = pd.DataFrame(data1)
df2 = pd.DataFrame(data2)
df3 = pd.DataFrame(data3)
```

```python
# DataFrame'leri birleştirme
combined_df = pd.concat([df1, df2, df3], ignore_index=True)
```

```python
# CSV dosyasına yazma
combined_df.to_csv("combined_data.csv", index=False)
```
